```python


import pandas as pd
import kagglehub
import os

# Download dataset
path = kagglehub.dataset_download("anvitkumar/shopping-dataset")

print("Dataset Path:", path)

# Show files inside dataset folder
print("\nFiles in dataset folder:")
print(os.listdir(path))
```

```python
csv_file = os.path.join(path, 'bra.csv')

df = pd.read_csv(csv_file)
```

```python
df.head()
```

```python
csv_file = os.path.join(path, 'ring.csv')

df = pd.read_csv(csv_file)
```

```python
print(df.isnull().sum())
```

```python
df.fillna(0, inplace=True)
```

```python
print(df.isnull().sum()) ## null and na values removed
```

```python
df.drop_duplicates(inplace=True)
```

```python
filtered_df = df[df['rating'] > 4]
```

```python
print("\nFiltered Rows:")
filtered_df[['title', 'rating' , 'product_description', 'final_price']]
```

```python
selected_columns = df[['title', 'rating', 'final_price','seller_name','discount']]

print("\nSelected Columns:")
print(selected_columns.head())
```

```python
df['final_price'] = (
    df['final_price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)
```

```python
df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')
df['ratings_count'] = pd.to_numeric(df['ratings_count'], errors='coerce')
```

```python
df['total_amount'] = df['final_price'] * df['ratings_count']
```

```python
print("Derived Column ")
print(df[['title', 'total_amount']])
```

```python
df.to_csv("cleaned_ring_dataset.csv", index=False)
```
